```python
import qrcode
import os


def generate_qr(text, filename, size=10, border=4):
    """
    Generate and save a QR code.
    """

    qr = qrcode.QRCode(
        version=None,
        error_correction=qrcode.constants.ERROR_CORRECT_M,
        box_size=size,
        border=border
    )

    qr.add_data(text)
    qr.make(fit=True)

    img = qr.make_image(
        fill_color="black",
        back_color="white"
    )

    # Add .png extension automatically
    if not filename.lower().endswith(".png"):
        filename += ".png"

    img.save(filename)

    return filename


def main():

    print("=" * 40)
    print("       QR CODE GENERATOR")
    print("=" * 40)

    # Get input
    text = input("\nEnter text or URL: ").strip()

    # Validate input
    if not text:
        print("Error: Text or URL cannot be empty!")
        return

    # Get filename
    filename = input("Enter file name: ").strip()

    if not filename:
        filename = "qrcode"

    # Get QR size
    while True:
        try:
            size = int(input("Enter QR size (5-15): "))

            if 5 <= size <= 15:
                break

            print("Please enter a number between 5 and 15.")

        except ValueError:
            print("Please enter a valid number.")

    # Generate QR
    try:

        output_file = generate_qr(
            text,
            filename,
            size
        )

        print("\n" + "-" * 40)
        print("QR Code generated successfully!")
        print(f"Saved as : {output_file}")
        print("-" * 40)

    except Exception as e:
        print("Error while generating QR code:", e)


# Program starts here
if __name__ == "__main__":
    main()
```
